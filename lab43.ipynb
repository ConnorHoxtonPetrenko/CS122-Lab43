Not exist x, x > 10, x {12, 3, 5, 7, 15}

````python
pnumbers = []
for N in range(10, 21):
    for i in range(2, N):
        if N % i == 0:
            break
    else:
        pnumbers.append(N)

print(pnumbers)

for v in pnumbers:
    print(v, end=' ')
````
